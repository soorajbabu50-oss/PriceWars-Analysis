```python
# -------------------------------
# 📘 1. Imports and Setup
# -------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="muted")

zomato = pd.read_csv("../data/processed/zomato_clean.csv")
swiggy = pd.read_csv("../data/processed/swiggy_clean.csv")

print("Zomato:", zomato.shape)
print("Swiggy:", swiggy.shape)

# -------------------------------
# 📊 2. Basic Overview
# -------------------------------
display(zomato.head())
display(swiggy.head())

# Summary stats
zomato.describe(include='all').T
swiggy.describe(include='all').T

# Missing values check
print("Zomato missing:\n", zomato.isna().sum())
print("Swiggy missing:\n", swiggy.isna().sum())

# -------------------------------
# 📈 3. Rating Distribution
# -------------------------------
plt.figure(figsize=(8,5))
sns.histplot(zomato['rate_clean'], bins=20, color='orange', label='Zomato', kde=True)
sns.histplot(swiggy['rating_clean'], bins=20, color='blue', label='Swiggy', kde=True)
plt.legend()
plt.title("Restaurant Rating Distribution")
plt.xlabel("Rating")
plt.show()

# -------------------------------
# 💰 4. Price Comparison
# -------------------------------
plt.figure(figsize=(8,5))
sns.boxplot(x=['Zomato']*len(zomato), y=zomato['cost_two_clean'], color='orange')
sns.boxplot(x=['Swiggy']*len(swiggy), y=swiggy['cost_two_clean'], color='blue')
plt.title("Cost for Two: Zomato vs Swiggy")
plt.ylabel("Cost (₹)")
plt.show()

# -------------------------------
# 🥘 5. Top Cuisines
# -------------------------------
from collections import Counter

z_cuisines = Counter([c.strip() for sub in zomato['cuisines'].dropna() for c in sub.split(',')])
s_cuisines = Counter([c.strip() for sub in swiggy['cuisines_clean'].dropna() for c in sub.split(',')])

z_df = pd.DataFrame(z_cuisines.most_common(10), columns=['Cuisine', 'Count'])
s_df = pd.DataFrame(s_cuisines.most_common(10), columns=['Cuisine', 'Count'])

fig, ax = plt.subplots(1,2, figsize=(14,5))
sns.barplot(x='Count', y='Cuisine', data=z_df, ax=ax[0], color='orange')
sns.barplot(x='Count', y='Cuisine', data=s_df, ax=ax[1], color='blue')
ax[0].set_title("Top 10 Cuisines - Zomato")
ax[1].set_title("Top 10 Cuisines - Swiggy")
plt.show()

# -------------------------------
# ⭐ 6. Rating vs Cost
# -------------------------------
plt.figure(figsize=(8,5))
sns.scatterplot(x='cost_two_clean', y='rate_clean', data=zomato, alpha=0.4, label='Zomato', color='orange')
sns.scatterplot(x='cost_two_clean', y='rating_clean', data=swiggy, alpha=0.4, label='Swiggy', color='blue')
plt.title("Cost vs Rating Comparison")
plt.xlabel("Cost for Two (₹)")
plt.ylabel("Rating")
plt.legend()
plt.show()

# -------------------------------
# 📍 7. Location Insights
# -------------------------------
plt.figure(figsize=(10,6))
z_toploc = zomato['location'].value_counts().head(10)
sns.barplot(x=z_toploc.values, y=z_toploc.index, color='orange')
plt.title("Top 10 Localities by Restaurant Count (Zomato - Bangalore)")
plt.show()

# -------------------------------
# 🧩 8. Platform Summary
# -------------------------------
summary = pd.DataFrame({
    'Platform': ['Zomato', 'Swiggy'],
    'Total Restaurants': [len(zomato), len(swiggy)],
    'Avg Rating': [zomato['rate_clean'].mean(), swiggy['rating_clean'].mean()],
    'Avg Cost for Two': [zomato['cost_two_clean'].mean(), swiggy['cost_two_clean'].mean()]
})
display(summary)
```
